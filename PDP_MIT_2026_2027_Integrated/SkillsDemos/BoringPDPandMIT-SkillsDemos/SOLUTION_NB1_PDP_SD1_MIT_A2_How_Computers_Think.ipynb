## Part 1: Warming Up with Binary (4 marks)Computers store everything as sequences of ones and zeros. Before we get into the interesting stuff, let's make sure we are comfortable converting between the number systems computers use and the ones we use.Here is a small puzzle to start: the number 42 in binary is `101010`. That is satisfying, isn't it? Let's figure out why.In decimal, `42` means 4 tens and 2 ones. In binary, `101010` means:`1x32 + 0x16 + 1x8 + 0x4 + 1x2 + 0x1 = 32 + 8 + 2 = 42`Each position is a power of 2 instead of a power of 10.

### Your turnWrite a function called `to_binary` that takes a positive integer and returns a string of ones and zeros representing that number in binary. There are many ways to do this -- one approach is to repeatedly divide by 2 and collect the remainders.Then write its companion, `from_binary`, which takes a binary string and returns the decimal integer.Think about what should happen with edge cases: what about 0? What about a binary string like `"0001"`?

```python
# Decimal to binary: repeatedly divide by 2 and record the remainder each
# time. Reading the remainders from last to first gives the binary digits.
# Special case: 0 itself needs to produce "0" rather than an empty string.
#
# Binary to decimal: walk the string left to right. For each character,
# double the running total so far and add the new digit (this is the same
# as adding digit * 2**power for each position, just accumulated as we go).
```

```python
def to_binary(n):
    """Convert a non-negative integer to a binary string."""
    if n == 0:
        return "0"
    digits = []
    while n > 0:
        digits.append(str(n % 2))
        n //= 2
    digits.reverse()
    return "".join(digits)
```

```python
def from_binary(binary_str):
    """Convert a binary string to its decimal integer value."""
    value = 0
    for digit in binary_str:
        value = value * 2 + int(digit)
    return value
```

### Testing your convertersTry a few conversions and verify they work. A good test: converting to binary and back should give you the original number.

```python
# Test your functions here
# Some interesting numbers to try: 0, 1, 7, 8, 42, 255, 256
for n in [0, 1, 7, 8, 42, 255, 256]:
    b = to_binary(n)
    back = from_binary(b)
    print(f"{n} -> {b} -> {back}  (round trip matches: {back == n})")

# A leading zero shouldn't confuse from_binary
print(from_binary("0001"), "should be 1")
```

Hexadecimal is another system computers use, where each digit represents 4 binary digits. The digits go 0-9 then A-F (so A=10, B=11, ... F=15). Can you write a `to_hex` function? If you get stuck, just describe the approach you would take -- that reasoning matters.

```python
def to_hex(n):
    """Convert a non-negative integer to an uppercase hexadecimal string."""
    if n == 0:
        return "0"
    hex_digits = "0123456789ABCDEF"
    digits = []
    while n > 0:
        digits.append(hex_digits[n % 16])
        n //= 16
    digits.reverse()
    return "".join(digits)

# Quick check against Python's own hex()
for n in [0, 15, 16, 42, 255, 4096]:
    print(n, to_hex(n), hex(n))
```

## Part 2: Lists as the Building Blocks of Everything (8 marks)Now that we have warmed up, let's talk about lists. In mathematics, we might call an ordered collection of numbers a *sequence* or a *vector*. In Python, we call it a list. But the idea is the same: a collection of items where order matters and we can access any item by its position.Mathematicians have a lovely notation for working with sequences. If we have a list of numbers `a = [3, 7, 2, 9, 1]`, we can write the sum of all elements as:$$\sum_{i=0}^{4} a_i = 3 + 7 + 2 + 9 + 1 = 22$$And the product of all elements as:$$\prod_{i=0}^{4} a_i = 3 \times 7 \times 2 \times 9 \times 1 = 378$$That capital sigma (sum) and capital pi (product) are just compact ways of saying "loop through and accumulate." Let's make that connection explicit.

### Your turnWrite two functions:1. `sigma(numbers)` -- takes a list of numbers and returns their sum. Do not use Python's built-in `sum()` -- write the loop yourself so you can feel the sigma notation in action.2. `pi_product(numbers)` -- takes a list of numbers and returns their product. Same rule: write the loop.Then write a third function, `dot_product(a, b)`, which takes two lists of the same length and computes their dot product: $\sum_{i=0}^{n-1} a_i \times b_i$. This one combines both ideas.

```python
def sigma(numbers):
    """Return the sum of a list of numbers."""
    total = 0
    for number in numbers:
        total += number
    return total
```

```python
def pi_product(numbers):
    """Return the product of a list of numbers."""
    product = 1
    for number in numbers:
        product *= number
    return product
```

```python
def dot_product(a, b):
    """Return the dot product of two equal-length lists."""
    # What should happen if the lists are different lengths?
    # Nothing sensible -- there is no way to pair every element up, so we
    # raise an error rather than silently truncating one of the lists.
    if len(a) != len(b):
        raise ValueError("dot_product needs two lists of the same length")
    total = 0
    for i in range(len(a)):
        total += a[i] * b[i]
    return total
```

```python
# Test all three functions
numbers = [3, 7, 2, 9, 1]
print("sigma:", sigma(numbers), "expected 22")
print("pi_product:", pi_product(numbers), "expected 378")
print("dot_product:", dot_product([1, 2, 3], [4, 5, 6]), "expected 32")
```

## Part 3: Finding Things -- Search Algorithms (8 marks)Suppose we have a list of student ID numbers and we need to find whether a particular ID is in the list. The simplest approach is to look at every single element until we find it (or run out of list). This is called *linear search* and it is the algorithm you use when you scan a shelf for a particular book.Let's build it, and then discover why it sometimes isn't good enough.

### Your turnWrite a function `linear_search(items, target)` that takes a list and a target value, and returns the index where the target was found, or -1 if it was not found.Then ask yourself: if the list has 1000 items, how many comparisons might we need in the worst case? What about 1,000,000 items?

```python
def linear_search(items, target):
    """Return the index of target in items, or -1 if it is not present."""
    for i in range(len(items)):
        if items[i] == target:
            return i
    return -1
```

```python
# Test it with a few cases, including one where the target is not in the list
print(linear_search([5, 3, 8, 1, 9], 8), "expected 2")
print(linear_search([5, 3, 8, 1, 9], 100), "expected -1")
print(linear_search([], 4), "expected -1")
```

### A faster way: Binary SearchHere is an idea: what if the list were already sorted? Then we could look at the middle element and immediately rule out half the list. This is *binary search*, and it is spectacularly efficient.Think about the "guess a number between 1 and 100" game. If someone tells you "higher" or "lower" after each guess, you would start at 50, then go to 25 or 75, and so on. You can always find the number in at most 7 guesses. That is binary search.The key insight is that each step *halves* the remaining possibilities. For a list of 1,000,000 items, binary search needs at most about 20 comparisons. Linear search might need 1,000,000.

### Your turnWrite a function `binary_search(sorted_items, target)` that takes a *sorted* list and a target, and returns the index where the target was found, or -1 if not found.The approach: keep track of a `low` and `high` index. Look at the middle. If the middle element is the target, we are done. If the target is smaller, search the left half. If larger, search the right half. Repeat until found or `low > high`.Before coding, write out the steps in plain English or pseudocode. What happens when the list is empty? What about a list with one element?

```python
# low = 0, high = len(items) - 1
# while low <= high:
#     mid = (low + high) // 2
#     if items[mid] == target: found it, return mid
#     elif items[mid] < target: the target must be in the right half, so low = mid + 1
#     else: the target must be in the left half, so high = mid - 1
# If the loop ends without finding it, return -1.
#
# Empty list: low starts at 0, high at -1, the loop body never runs, we return -1.
# Single-element list: mid is that element, so one comparison settles it either way.
```

```python
def binary_search(sorted_items, target):
    """Return the index of target in a sorted list, or -1 if not found."""
    low = 0
    high = len(sorted_items) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_items[mid] == target:
            return mid
        elif sorted_items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Test it -- make sure to test with sorted lists!
sorted_numbers = [1, 3, 4, 7, 9, 12, 15, 20]
print(binary_search(sorted_numbers, 9), "expected 4")
print(binary_search(sorted_numbers, 5), "expected -1")
print(binary_search([], 5), "expected -1")
print(binary_search([42], 42), "expected 0")

# What happens if you accidentally give it an unsorted list?
# The halving logic assumes the list is sorted, so it can throw away the
# half that actually contains the target -- the search can quietly return
# the wrong answer, or -1 for a value that is really there.
print(binary_search([9, 1, 4, 20, 3], 3), "unsorted input -- may well be wrong")
```

## Part 4: Putting Things in Order -- Sorting Algorithms (10 marks)Binary search needs a sorted list. So how do we sort a list? This turns out to be one of the most studied problems in computer science, and there are many different approaches. Let's explore a few.

### Bubble SortThe simplest sorting algorithm works like this: walk through the list comparing adjacent pairs. If they are in the wrong order, swap them. Repeat until no more swaps are needed.It is called bubble sort because the largest values "bubble up" to the end of the list, one pass at a time.

### Your turnWrite a function `bubble_sort(items)` that sorts a list in ascending order using bubble sort. Return the sorted list.A few things to think about:- How do you swap two elements in a list?- How do you know when to stop?- Can you count how many comparisons your function makes?

```python
# Walk through the list from the start, comparing each pair of neighbours.
# If the left one is bigger than the right one, swap them. After one full
# pass, the largest value has "bubbled" to the end, so the next pass only
# needs to look at one fewer position. Stop once a whole pass makes no
# swaps at all -- that is the signal that the list is already sorted.
```

```python
def bubble_sort(items):
    """Sort a list in ascending order using bubble sort."""
    items = items.copy()
    n = len(items)
    for i in range(n):
        swapped = False
        for j in range(n - 1 - i):
            if items[j] > items[j + 1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                swapped = True
        if not swapped:
            break
    return items
```

```python
# Test it with several lists: already sorted, reverse sorted, duplicates, empty
print(bubble_sort([5, 3, 8, 1, 9]))
print(bubble_sort([1, 2, 3, 4, 5]))
print(bubble_sort([5, 4, 3, 2, 1]))
print(bubble_sort([3, 1, 3, 2, 1]))
print(bubble_sort([]))
```

### Insertion SortHere is another approach, one that is closer to how you might sort a hand of cards. Start with the first card. Pick up the next card and insert it into the correct position among the cards you have already sorted. Repeat.

### Your turnWrite `insertion_sort(items)`. Again, think through the logic before coding. Where does each new element get inserted?

```python
def insertion_sort(items):
    """Sort a list in ascending order using insertion sort."""
    items = items.copy()
    for i in range(1, len(items)):
        current = items[i]
        j = i - 1
        while j >= 0 and items[j] > current:
            items[j + 1] = items[j]
            j -= 1
        items[j + 1] = current
    return items
```

```python
# Test it
print(insertion_sort([5, 3, 8, 1, 9]))
print(insertion_sort([1, 2, 3, 4, 5]))
print(insertion_sort([5, 4, 3, 2, 1]))
print(insertion_sort([]))
```

### Selection SortOne more approach: find the smallest element and put it first. Then find the smallest element in the remainder and put it second. And so on.

### Your turnWrite `selection_sort(items)`. This one has a nice structure: an outer loop that decides which position to fill, and an inner loop that finds the minimum in the remaining unsorted portion.

```python
def selection_sort(items):
    """Sort a list in ascending order using selection sort."""
    items = items.copy()
    n = len(items)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if items[j] < items[min_index]:
                min_index = j
        items[i], items[min_index] = items[min_index], items[i]
    return items
```

```python
# Test it
print(selection_sort([5, 3, 8, 1, 9]))
print(selection_sort([1, 2, 3, 4, 5]))
print(selection_sort([5, 4, 3, 2, 1]))
print(selection_sort([]))
```

```python
# Optional: recursive binary search and/or shell sort
def binary_search_recursive(sorted_items, target, low=0, high=None):
    """Recursive version of binary search."""
    if high is None:
        high = len(sorted_items) - 1
    if low > high:
        return -1
    mid = (low + high) // 2
    if sorted_items[mid] == target:
        return mid
    elif sorted_items[mid] < target:
        return binary_search_recursive(sorted_items, target, mid + 1, high)
    else:
        return binary_search_recursive(sorted_items, target, low, mid - 1)


print(binary_search_recursive(sorted_numbers, 9), "expected 4")
print(binary_search_recursive(sorted_numbers, 5), "expected -1")


def shell_sort(items):
    """Sort a list in ascending order using shell sort."""
    items = items.copy()
    n = len(items)
    gap = n // 2
    while gap > 0:
        for i in range(gap, n):
            current = items[i]
            j = i
            while j >= gap and items[j - gap] > current:
                items[j] = items[j - gap]
                j -= gap
            items[j] = current
        gap //= 2
    return items


print(shell_sort([5, 3, 8, 1, 9, 2, 7, 4, 6, 0]))

# Shell sort moves elements long distances early on, instead of one step at
# a time like plain insertion sort, so values that start far from their
# final position settle into roughly the right neighbourhood much sooner.
```
