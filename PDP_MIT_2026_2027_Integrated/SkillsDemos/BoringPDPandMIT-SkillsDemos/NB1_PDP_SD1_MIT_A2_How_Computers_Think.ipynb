## Part 2: Lists as the Building Blocks of Everything (8 marks)Now that we have warmed up, let's talk about lists. If we have a list of numbers `a = [3, 7, 2, 9, 1]`, we can write the sum of all elements as:$$\sum_{i=0}^{4} a_i = 3 + 7 + 2 + 9 + 1 = 22$$And the product of all elements as:$$\prod_{i=0}^{4} a_i = 3 \times 7 \times 2 \times 9 \times 1 = 378$$That capital sigma (sum) and capital pi (product) are just compact ways of saying "loop through and accumulate." Let's make that connection explicit.

### A faster way: Binary SearchHere is an idea: what if the list were already sorted? Then we could look at the middle element and immediately rule out half the list. This is *binary search*, and it is spectacularly efficient.Think about the "guess a number between 1 and 100" game. If someone tells you "higher" or "lower" after each guess, you would start at 50, then go to 25 or 75, and so on. You can always find the number in at most 7 guesses. That is binary search.The key insight is that each step *halves* the remaining possibilities. For a list of 1,000,000 items, binary search needs at most about 20 comparisons. Linear search might need 1,000,000.
