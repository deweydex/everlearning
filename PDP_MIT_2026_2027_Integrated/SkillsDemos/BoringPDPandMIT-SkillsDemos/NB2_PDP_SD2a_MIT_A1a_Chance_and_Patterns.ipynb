### Permutations and CombinationsA *permutation* counts the number of ways to arrange r objects chosen from n distinct objects, where order matters. The formula is:$$P(n, r) = \frac{n!}{(n-r)!}$$A *combination* counts the number of ways to choose r objects from n where order does not matter:$$C(n, r) = \frac{n!}{r! \cdot (n-r)!}$$The difference: choosing 3 books from a shelf of 10 to read in order (permutation) versus choosing 3 books to pack in a bag (combination).

## Part 2: Probability (8 marks)Probability is a number between 0 and 1 that measures how likely something is to happen. An impossible event has probability 0. A certain event has probability 1. Everything else falls somewhere in between.The basic idea: if all outcomes are equally likely, the probability of an event is the number of favourable outcomes divided by the total number of outcomes.$$P(A) = \frac{\text{number of outcomes where A happens}}{\text{total number of outcomes}}$$

For mutually exclusive events: $P(A \text{ or } B) = P(A) + P(B)$.For events that are not mutually exclusive: $P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)$.

### Central tendencyThe three classic measures of the "centre" of a dataset:- **Mean**: the arithmetic average. Sum everything and divide by the count.- **Median**: the middle value when the data is sorted. (If there is an even number of values, take the average of the two middle ones.)- **Mode**: the value that appears most frequently.Each tells us something different. The mean is sensitive to extreme values (outliers). The median is robust to them. The mode tells us what is most common but might not exist uniquely.

### Spread: Range and Standard DeviationKnowing the centre is not enough. Two classes could have the same mean but very different spreads: one where everyone scored around 38, and another where scores ranged from 10 to 50.The **range** is the simplest measure: maximum minus minimum.The **standard deviation** is more informative. It measures how far, on average, each data point is from the mean:$$\sigma = \sqrt{\frac{\sum_{i=1}^{N}(x_i - \mu)^2}{N}}$$where $\mu$ is the mean and $N$ is the number of data points.
