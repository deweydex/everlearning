To start let's look at the python random library. In particular lets look at the documentation [here](https://docs.python.org/3/library/random.html). With that source in mind, lets see if we can create a function that shuffles our deck and returns a card. For this we will use an optional argument `num_cards` which will default to 1.

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** `random.shuffle()` shuffles a list in place, so I shuffle a copy of the deck rather than the original, then slice off the first `num_cards` items as the "picked" cards and the rest as what remains. Slicing is convenient here because a negative or zero `num_cards` gives an empty slice-shaped answer, but I still check for `num_cards <= 0` explicitly so the function's behaviour reads clearly rather than relying on that Python quirk.


```python
def pick_random_card(deck, num_cards=1):
    """Shuffle a copy of deck and pull the top num_cards cards off it.
    Returns (picked_cards, remaining_deck)."""
    if num_cards <= 0:
        # nothing to pick, so hand back an empty hand and the untouched deck
        return [], deck
    shuffled = deck.copy()
    random.shuffle(shuffled)
    picked = shuffled[:num_cards]
    remaining = shuffled[num_cards:]
    return picked, remaining
```

```python
# go ahead and create your function in this cell!

# Bug fix notes: the starter code converted the input to an int and checked
# "if card in deck", but our deck holds strings like "2S", so that check could never
# succeed. It also returned the untouched deck instead of removing the picked card.
def user_picks_a_card(deck):
    while True:
        card = input("Pick a card, any card... but make it a good one! \n" + str(deck) + "\n")
        if card in deck:
            remaining_deck = deck.copy()
            remaining_deck.remove(card)
            return card, remaining_deck
        else:
            print("that card is not in the deck, please try again")
```

### A flip of the previous function
What if we made a function `computer_guesses_card(users_card, deck)` that would guess the users card with one of our prior functions and then returns how many guesses it took.

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** Rather than write new picking logic, I reuse `pick_random_card` from earlier in a loop: the computer keeps drawing a random card from its own copy of the deck, counting attempts, until it draws one equal to `users_card`. I hand the loop a fresh, full copy of the deck to guess from so its search doesn't run out of cards before it finds a match.


```python
# Write your function here!
def computer_guesses_card(users_card, deck):
    """Have the computer guess users_card by repeatedly drawing a random card from
    its own copy of deck, counting how many guesses it takes."""
    guess_pool = deck.copy()
    attempts = 0
    guess = None
    while guess != users_card:
        if not guess_pool:
            # ran through every card without a match (shouldn't normally happen if
            # users_card is really in deck) -- reshuffle a fresh pool and keep trying
            guess_pool = deck.copy()
        picked, guess_pool = pick_random_card(guess_pool, 1)
        guess = picked[0]
        attempts += 1
    return attempts
```

Let's start out with a couple smaller functions: can you create a function that takes a card (string) and a guess (string) as input that checks whether the guess is a color and returns whether the guess is true?

```python
def is_right_color(card: str, guess: str) -> bool:
    """card is a string like "2S" or "10D"; guess is "Red" or "Black"."""
    red_suits = {"H", "D"}
    black_suits = {"S", "C"}
    suit = card[-1]
    guess_clean = guess.strip().lower()
    if guess_clean == "red":
        return suit in red_suits
    elif guess_clean == "black":
        return suit in black_suits
    return False
```

```python
def is_right_suit(card: str, guess: str) -> bool:
    """card is a string like "2S" or "10D"; guess is a suit letter like "S" or "d"."""
    suit = card[-1]
    return guess.strip().upper() == suit
```

### Building a key piece of game logic:
Let's try to create a function called `lower_higher()` that takes a `card` (string) and a `guess` (string) as input that indicates both a direction (>, =, <) and a value checks whether the guess meets that criteria and returns whether the guess is true? (for example the guess `>4` would return `true` for `6` or `K` but `false` for `3` or `2`.) Note, there are many ways of doing this, the shorter "character + number" approach is just one way, feel free to try something else!

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** I split the guess into a direction character (`>`, `<`, or `=`) and a value, using the same rank logic as `is_right_value`, but I also need a way to compare face cards to numbers, so I map `J`, `Q`, `K`, `A` to 11 through 14 first. Once both the card and the guessed value have a numeric rank, the comparison itself is a single if/elif/else on the direction character.


```python
def _card_rank(value: str) -> int:
    """Turn a card value like "2", "10", or "K" into a comparable integer rank."""
    face_ranks = {"J": 11, "Q": 12, "K": 13, "A": 14}
    value = value.strip().upper()
    if value in face_ranks:
        return face_ranks[value]
    return int(value)

def lower_higher(card: str, guess: str) -> bool:
    """guess looks like ">4", "<7", or "=K": a direction plus a value to compare card to."""
    direction = guess[0]
    target_value = guess[1:]
    card_rank = _card_rank(card[:-1])
    target_rank = _card_rank(target_value)
    if direction == ">":
        return card_rank > target_rank
    elif direction == "<":
        return card_rank < target_rank
    elif direction == "=":
        return card_rank == target_rank
    return False
```

```python
# Create some test cases
print(lower_higher("6S", ">4"))  # True
print(lower_higher("3S", ">4"))  # False
print(lower_higher("KS", ">4"))  # True: King outranks 4
print(lower_higher("2S", "<3"))  # True
print(lower_higher("7H", "=7"))  # True

# use your function for this challenge on your test cases to see if it works as you intend
```

```python
def handle_user_guess(card: str) -> bool:
    """Ask the player what kind of guess they'd like to make, then run the matching
    checker function on it."""
    guess = input(
        "What's your guess? A color (Red/Black), a suit (S/H/C/D), a value "
        "(2-10, J, Q, K, A), or a direction like >4, <7, or =K.\n"
    )
    guess_clean = guess.strip()
    if guess_clean.lower() in ("red", "black"):
        return is_right_color(card, guess_clean)
    elif guess_clean.upper() in ("S", "H", "C", "D"):
        return is_right_suit(card, guess_clean)
    elif guess_clean and guess_clean[0] in ("<", ">", "="):
        return lower_higher(card, guess_clean)
    else:
        return is_right_value(card, guess_clean)

# quick check with a simulated player input
with patch('builtins.input', side_effect=["Red"]):
    print(handle_user_guess("KH"))  # True
```

Finally, let's create a function that houses our core gameplay logic. But let's do that as a group!
```python
# contains a loop
# asks the player if they would like to play (if yes)
# leads to another loop
# computer picks a card
# display options the user can input
# user guesses with those rules for input
# this continues, until user finds card

def is_right_color(card: str, guess: str) -> bool:
    red_suits = {"H", "D"}
    black_suits = {"S", "C"}
    suit = card[-1]
    guess_clean = guess.strip().lower()
    if guess_clean == "red":
        return suit in red_suits
    elif guess_clean == "black":
        return suit in black_suits
    return False

def is_right_suit(card : str, guess : str) -> bool:
    return guess.strip().upper() == card[-1]

def is_right_value(card: str, guess: str) -> bool:
    return guess.strip().upper() == card[:-1].upper()

def lower_higher(card:str, guess:str)-> bool:
    direction = guess[0]
    card_rank = _card_rank(card[:-1])
    target_rank = _card_rank(guess[1:])
    if direction == ">":
        return card_rank > target_rank
    elif direction == "<":
        return card_rank < target_rank
    elif direction == "=":
        return card_rank == target_rank
    return False

# we have a way to pick a card from task two!

def display_possible_strings():
    print("You can guess in a few different ways:")
    print("  - a color: Red or Black")
    print("  - a suit: S, H, C, or D")
    print("  - a value: 2 through 10, J, Q, K, or A")
    print("  - a direction and value, like >4, <7, or =K")
    print("  - or the whole card, like 'KH', to make a final guess")
    print("Type 'quit' at any point to leave the game.")

def gameplay_loop():
    display_possible_strings()
    computer_card, _ = pick_random_card(deck_strings, 1)
    computer_card = computer_card[0]
    attempts = 0
    while True:
        guess = input("Ask a question, or guess the whole card to win. Type 'quit' to leave.\n")
        guess = guess.strip()
        if guess.lower() == "quit":
            print("Thanks for playing!")
            return None
        attempts += 1
        if guess.upper() == computer_card:
            print(f"That's it! The card was {computer_card}.")
            return attempts
        elif guess.lower() in ("red", "black"):
            print("Yes!" if is_right_color(computer_card, guess) else "No.")
        elif guess.upper() in ("S", "H", "C", "D"):
            print("Yes!" if is_right_suit(computer_card, guess) else "No.")
        elif guess and guess[0] in "<>=":
            print("Yes!" if lower_higher(computer_card, guess) else "No.")
        else:
            print("Yes!" if is_right_value(computer_card, guess) else "No.")

def guessing_game():
    want_to_play = input("Would you like to play a guessing game? y/n")
    if want_to_play.strip().lower() == "y":
        result = gameplay_loop()
        if result is not None:
            print(f"it took you {result} number of guesses to guess my card!")

    else:
        print("okay bye!")
```

### Sample solution for the Bonus Task (optional, not required)

One way to guess in at most 6 tries: 52 cards fit inside 6 yes/no questions because 2**6 = 64 is greater than 52. Sort the deck into one fixed order and binary search it: each guess should roughly halve how many cards are still possible.

```python
def _card_sort_key(card):
    """Turn a card string into a single sortable integer 0-51: rank first, suit second."""
    suit_order = {"S": 0, "H": 1, "C": 2, "D": 3}
    return _card_rank(card[:-1]) * 4 + suit_order[card[-1]]

def computer_guesses_optimally(users_card, deck):
    """Binary search a sorted, de-duplicated deck for users_card, returning how
    many guesses it took."""
    sorted_deck = sorted(set(deck), key=_card_sort_key)
    low, high = 0, len(sorted_deck) - 1
    target_key = _card_sort_key(users_card)
    attempts = 0
    while low <= high:
        mid = (low + high) // 2
        attempts += 1
        guess_key = _card_sort_key(sorted_deck[mid])
        if guess_key == target_key:
            return attempts
        elif guess_key < target_key:
            low = mid + 1
        else:
            high = mid - 1
    return attempts

print(computer_guesses_optimally("KH", deck_strings))

# confirm it never needs more than 6 guesses for any card in the deck
worst_case = max(computer_guesses_optimally(card, deck_strings) for card in set(deck_strings))
print(f"worst case across the whole deck: {worst_case} guesses")
assert worst_case <= 6
```
