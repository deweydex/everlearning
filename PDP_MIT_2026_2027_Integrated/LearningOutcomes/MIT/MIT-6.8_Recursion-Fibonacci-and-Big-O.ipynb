## 1. Slow Fibonacci
This one uses recursion and repeats the same operation again and again. It is simple but inefficient.

```python
def slowfib(n):
    if n == 0 or n == 1:
        return 1
    return slowfib(n - 1) + slowfib(n - 2)

print("slowfib(5):", slowfib(5))
```

**Time complexity:** Extremely slow time complexity: O(2^n)

It does an lot of duplicate work.

## 2. Faster Fibonacci
This one remembers answers it already found, so it doesn’t repeat the same work.

```python
def fastfib(n, prev):
    if n == 0 or n == 1:
        return 1
    if prev[n] != 0:
        return prev[n]
    prev[n] = fastfib(n-1, prev) + fastfib(n-2, prev)
    return prev[n]

n = 10
memo = [0] * (n + 1)
print("fastfib(10):", fastfib(n, memo))
```

**Time complexity:** A lot quicker → O(n)

We keep past results so we don't repeat.

**Trade-off:** We use more memory.
This is called "memoization".

## 3. Iterative Fibonaci
This version just uses a loop and no recursion.

```python
def fasterfib(n):
    if n == 0 or n == 1:
        return 1
    a, b = 1, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

print("fasterfib(10):", fasterfib(10))
```

**Time complexity:** O(n)  
This is the fastest and most elegant in pure Python.
No additional memory required apart from 2 variables.
