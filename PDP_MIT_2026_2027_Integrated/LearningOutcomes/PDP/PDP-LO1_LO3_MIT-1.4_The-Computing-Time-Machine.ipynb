## Stop 1: 1843 — Ada Lovelace and the Analytical Engine

Before there were computers, there were **mechanical calculating machines**. Charles Babbage designed the Analytical Engine — a general-purpose mechanical computer that was never fully built in his lifetime. But it was **Ada Lovelace**, translating an Italian paper about the machine, who added something Babbage hadn't: a set of notes describing how the Engine could be *programmed* to compute a sequence of numbers (Bernoulli numbers), using loops and conditional branching, entirely on paper.

That makes her, by most historians' account, **the first computer programmer** — over a century before there was an electronic computer to run her program on.

**Key idea:** a program doesn't need electricity, or even a machine that works yet, to exist. It's a sequence of precise instructions. Everything else in this notebook is about *how those instructions get carried out*.

## Stop 2: 1945 — Machine Code, the Only Language the Machine Understands

Early electronic computers like **ENIAC** (1945) had no programming language at all. To "program" ENIAC, engineers physically rewired it with plugboards and switches. A few years later, machines were built to read instructions from memory instead — but those instructions were still just patterns of **on/off**, which we write as binary (base 2): every instruction, every number, every piece of text, all of it, was binary.

This is **machine code** — the raw, native language of the hardware. No compiler, no interpreter, nothing standing between you and the circuits.

Binary works because a transistor (or a vacuum tube, in ENIAC's day) is naturally a two-state device: on or off, high voltage or low voltage. Base 2 maps onto that perfectly. Base 10 (what we use day to day) doesn't map onto hardware nearly as cleanly.

Run the cell below to load the binary conversion tools you'll need for this stop (adapted from `Tutorial_02`).

```python
def to_binary(n):
    """Convert a non-negative integer to a binary string (no '0b' prefix)."""
    if n == 0:
        return "0"
    digits = []
    while n > 0:
        digits.append(str(n % 2))
        n = n // 2
    return "".join(reversed(digits))


def from_binary(bstr):
    """Convert a binary string back to a base-10 integer."""
    total = 0
    for ch in bstr:
        total = total * 2 + int(ch)
    return total


# quick sanity check
print(to_binary(42))       # should be 101010
print(from_binary("101010"))  # should be 42
```

## Stop 3: 1950s — Assembly Language and Hexadecimal

Writing raw binary by hand is exhausting and extremely easy to get wrong — a string like `01001000` is hard for a human to read at a glance. Two things happened next:

1. **Assembly language** gave each machine instruction a short human-readable name (`ADD`, `MOV`, `JMP`) instead of a binary pattern. A program called an **assembler** translated assembly back into the binary machine code the hardware actually needed.
2. **Hexadecimal (base 16)** became the standard shorthand for writing binary, because *one hex digit exactly represents four binary digits* (a "nibble"). `1111` in binary is `F` in hex — much easier to read and write than a long string of 1s and 0s.

This is exactly why MIT 1.4 asks you to convert between binary, hex, and base 10 — hex only exists as a convenience for reading and writing binary.

```python
def to_hex(n):
    """Convert a non-negative integer to an uppercase hex string (no '0x' prefix)."""
    return format(n, "X")


def hex_to_binary(hexstr):
    """Convert a hex string to a binary string."""
    n = int(hexstr, 16)
    return to_binary(n)


# quick sanity check
print(to_hex(255))            # should be FF
print(hex_to_binary("FF"))    # should be 11111111
```

## Stop 4: 1957–1972 — High-Level Languages Take Over

Assembly was still tied to one specific machine's instruction set — a program written for one computer wouldn't run on another. The next leap was **high-level languages**: code that reads more like English or mathematics, and that gets translated to machine code by software rather than by a human assembler.

| Year | Language | What it was for |
|---|---|---|
| 1957 | **FORTRAN** | Scientific and engineering calculation |
| 1959 | **COBOL** | Business data processing |
| 1958–60 | **LISP** | Symbolic/mathematical reasoning — the ancestor of functional programming |
| 1972 | **C** | Systems programming (operating systems, close to the hardware) |
| 1991 | **Python** | General-purpose, readable, the language you're using right now! |

Two ways a high-level language gets turned into something the machine can run:

- A **compiler** translates the *entire* program into machine code (or another lower-level form) *before* it runs, producing a standalone executable. C is traditionally compiled.
- An **interpreter** reads and executes the program *line by line, as it runs*, with no separate translation step beforehand. Python is traditionally interpreted (technically it compiles to an intermediate "bytecode" first, then interprets that — a hybrid, but it behaves like an interpreted language from the programmer's point of view).

## Stop 5: Programming Paradigms — Same Problem, Different Styles

A **paradigm** is a fundamental style or approach to structuring a program. The same task — say, doubling every number in a list — looks quite different depending on the paradigm a language (or a programmer) leans on.

Run the cell below to see four different styles applied to the exact same problem, all in Python (Python happens to support all four styles, which is unusual and part of why it's such a popular teaching language).

```python
numbers = [1, 2, 3, 4, 5]

# --- Style A: procedural ---
# Step-by-step instructions that change state (the `doubled` list) as they go.
doubled_procedural = []
for n in numbers:
    doubled_procedural.append(n * 2)
print("Procedural:", doubled_procedural)

# --- Style B: functional ---
# Describe *what* transformation to apply, not *how* to loop.
doubled_functional = list(map(lambda n: n * 2, numbers))
print("Functional:", doubled_functional)

# --- Style C: object-oriented ---
# Bundle the data and the behaviour that acts on it together in a class.
class NumberList:
    def __init__(self, values):
        self.values = values

    def doubled(self):
        return [v * 2 for v in self.values]

print("Object-oriented:", NumberList(numbers).doubled())

# --- Style D: scripting one-liner ---
# Python's list comprehension: a compact, declarative style common in scripting languages.
doubled_scripting = [n * 2 for n in numbers]
print("Scripting/comprehension:", doubled_scripting)
```
