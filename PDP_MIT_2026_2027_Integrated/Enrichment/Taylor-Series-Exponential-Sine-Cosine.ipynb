## 1. Exponential Function

```python
import math

def factorial(n):
    return math.factorial(n)

def myexp(x, num_terms=None, threshold=None):
    result = 0
    n = 0
    while True:
        term = (x ** n) / factorial(n)
        result += term
        n += 1
        if num_terms and n >= num_terms:
            break
        if threshold and abs(term) < threshold:
            break
    return result

# Testing
print("Exponential of 1 (5 terms):", myexp(1, num_terms=5))
print("Exponential of 1 (threshold 0.001):", myexp(1, threshold=0.001))
```

## 2. Sine Function

```python
def mysine(x, num_terms=None, threshold=None):
    result = 0
    n = 0
    while True:
        term = ((-1) ** n) * (x ** (2 * n + 1)) / factorial(2 * n + 1)
        result += term
        n += 1
        if num_terms and n >= num_terms:
            break
        if threshold and abs(term) < threshold:
            break
    return result

# Testing
print("Sine of pi/2 (5 terms):", mysine(math.pi/2, num_terms=5))
print("Sine of pi/2 (threshold 0.001):", mysine(math.pi/2, threshold=0.001))
```

## 3. Cosine Function

```python
def mycosine(x, num_terms=None, threshold=None):
    result = 0
    n = 0
    while True:
        term = ((-1) ** n) * (x ** (2 * n)) / factorial(2 * n)
        result += term
        n += 1
        if num_terms and n >= num_terms:
            break
        if threshold and abs(term) < threshold:
            break
    return result

# Testing
print("Cosine of 0 (5 terms):", mycosine(0, num_terms=5))
print("Cosine of 0 (threshold 0.001):", mycosine(0, threshold=0.001))
```
