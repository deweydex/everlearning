**Sample response (not the only right way to answer this):** "Same length" just means comparing two counts, and Python already hands us that count with `len()`, so there is not much to invent here.
```python
# Break it down first: two lists are the "same length" if len(list1) equals len(list2).
# Python already gives us len() for that, so this function is mostly about naming the idea clearly.

def same_length(list1, list2):
    """Return True if list1 and list2 contain the same number of elements."""
    return len(list1) == len(list2)
```

# Challenge 2: 10%

See if you can write and test a function called `list_addition` that accepts two lists of integers as its arguments and returns a new list containing the sum of each entry in the first list with each entry in the second list. Be sure to check that the lists are of the same length! if they are not, return an empty list.

As an example list_addition([12, 34, 56], [-4, 91, 0]) would return [8, 125, 56].

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** Before writing any code, I checked the lengths match, using the `same_length` function from Challenge 1 rather than writing the same check twice. If they don't match, the instructions say to return an empty list rather than raise an error, so I return `[]` early and let the rest of the function assume the lists line up. Once I know the lists are the same length, `zip` pairs up the matching entries so I can add each pair with a list comprehension.


```python
# Remember to first break down your thinking using comment
# the more english you write here, the easier the code will be later!
# Plan: if the lists aren't the same length, return []. Otherwise, pair up entries
# with zip() and add each pair together into a new list.

def list_addition(list1, list2):
    """Add matching entries of two integer lists together. Returns [] if the lists
    are not the same length."""
    if not same_length(list1, list2):
        return []
    return [a + b for a, b in zip(list1, list2)]
```

# Challenge 3: 10%

Let's try to code a function that implements digit addition using lists. To do this, lets create a function `gradeschool_addition` that takes in two lists of the same length which contain only single digit positive integers as each list element. This function would return a new list (potentially a slightly longer list) which represents the digit (gradeschool) addition of the two lists: so adding 1234 + 2345 would give 3579 normally, but in our case we would have `gradeschool_addition([1,2,3,4],[2,3,4,5]) -> [3,5,7,9]`(see how the lists are basically numbers)?

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** This is the same idea as adding numbers on paper: start from the rightmost (ones) digit, add each pair, and carry a 1 into the next column whenever a pair sums to 10 or more. I walk the two lists from right to left with `reversed()`, track a running `carry`, and build up the result digit by digit. If there's a leftover carry once I've used every digit, that becomes a new, most-significant digit, which is why the result can end up one digit longer than the inputs.


```python
# Remember to first break down your thinking using comment
# the more english you write here, the easier the code will be later!
# Plan: walk both lists from the rightmost digit to the leftmost, adding each pair plus
# any carry from the previous column. If a column's total is 10 or more, keep the ones
# digit and carry a 1 into the next column. If there's a carry left after the last
# column, that becomes a new leading digit.

def gradeschool_addition(list1, list2):
    """Add two lists of single-digit integers (most significant digit first) the way
    you would add two numbers by hand, and return the digit-list result."""
    result_digits = []
    carry = 0
    for digit1, digit2 in zip(reversed(list1), reversed(list2)):
        total = digit1 + digit2 + carry
        result_digits.append(total % 10)
        carry = total // 10
    if carry:
        result_digits.append(carry)
    return list(reversed(result_digits))
```

# Challenge 4: 5%

Can you write a function that accepts lists of different length and handles them appropriately?

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** Numbers of different lengths still line up at the ones digit, not at the front of the list, so the fix is to pad the shorter list with leading zeros until both lists match in length, then reuse `gradeschool_addition` from Challenge 3 rather than duplicating that logic. Padding with zeros doesn't change the value of the number, only how many digits we're representing it with, so it's a safe way to make the two lists comparable.


```python
# Remember to first break down your thinking using comment
# the more english you write here, the easier the code will be later!
# Plan: work out how many digits shorter one list is than the other, pad the shorter
# list with that many leading zeros (leading zeros don't change a number's value),
# and then hand both equal-length lists off to gradeschool_addition from Challenge 3.

def gradeschool_addition_any_length(list1, list2):
    """Add two lists of single-digit integers even when they're different lengths,
    by padding the shorter one with leading zeros first."""
    length_difference = len(list1) - len(list2)
    if length_difference > 0:
        list2 = [0] * length_difference + list2
    elif length_difference < 0:
        list1 = [0] * (-length_difference) + list1
    return gradeschool_addition(list1, list2)
```

## Testing for Challenge 4:

```python
# Create some test cases
print(gradeschool_addition_any_length([1, 2, 3, 4], [5]))        # 1234 + 5 = 1239
print(gradeschool_addition_any_length([9], [9, 9, 9]))           # 9 + 999 = 1008
print(gradeschool_addition_any_length([], [1, 2, 3]))            # 0 + 123 = 123
print(gradeschool_addition_any_length([1, 2, 3], []))            # 123 + 0 = 123

# use your function for this challenge on your test cases to see if it works as you intend
```

# Bonus Challenge: Just for Fun (0%)

How about an equivalent function for subtraction--that one might take significant changes to your previous method!

### Double Click this cell to modify it and write your thought process and show the "story" of your code!

**Sample response (not the only right way to answer this):** Subtraction needed real changes rather than a small tweak to addition: I now have to figure out which number is bigger before I start (so I know whether the answer is positive or negative), and instead of carrying I'm borrowing, so a negative column result gets 10 added to it and takes a 1 away from the next column. I represent the sign separately from the digits rather than trying to force a minus sign into a list of digits.


```python
# Remember to first break down your thinking using comment
# the more english you write here, the easier the code will be later!
# Plan: pad the two lists to the same length like in Challenge 4. Compare them as
# numbers to work out which one is bigger, and swap them (remembering the swap makes
# the answer negative) so we always subtract the smaller from the larger. Then borrow
# instead of carry: if a column would go negative, add 10 and take 1 away from the
# next column. Finally strip any leading zeros left over from the subtraction.

def _digits_to_int(digits):
    return int("".join(str(d) for d in digits)) if digits else 0

def gradeschool_subtraction(list1, list2):
    """Subtract the number represented by list2 from the number represented by list1.
    Returns a tuple: (is_negative, result_digits)."""
    length_difference = len(list1) - len(list2)
    if length_difference > 0:
        list2 = [0] * length_difference + list2
    elif length_difference < 0:
        list1 = [0] * (-length_difference) + list1

    is_negative = _digits_to_int(list1) < _digits_to_int(list2)
    if is_negative:
        list1, list2 = list2, list1

    result_digits = []
    borrow = 0
    for digit1, digit2 in zip(reversed(list1), reversed(list2)):
        value = digit1 - digit2 - borrow
        if value < 0:
            value += 10
            borrow = 1
        else:
            borrow = 0
        result_digits.append(value)
    result_digits.reverse()

    while len(result_digits) > 1 and result_digits[0] == 0:
        result_digits.pop(0)

    return is_negative, result_digits
```

## Testing for Bonus Challenge

```python
# Create some test cases
print(gradeschool_subtraction([5, 3], [2, 1]))     # (False, [3, 2])   53 - 21 = 32
print(gradeschool_subtraction([2, 1], [5, 3]))     # (True, [3, 2])    21 - 53 = -32
print(gradeschool_subtraction([1, 0, 0], [1]))     # (False, [9, 9])   100 - 1 = 99
print(gradeschool_subtraction([4], [4]))           # (False, [0])      4 - 4 = 0

# use your function for this challenge on your test cases to see if it works as you intend
```
