## 2. Changing a value

A program often needs a number that changes: a score, a balance, a count. Run this cell after you write your guess for the last line.

````python
# My guess for the last line:
points = 10
points = points + 5
print(points)
````

The cell printed 15. Python follows two steps for the second line. First, it works out the right side: `points + 5` is 15. Second, it attaches the name `points` to that new value.

In a maths class, `points = points + 5` could never be true. In Python, the `=` sign is an instruction. It does not say that two things are equal. We read the line as *take the value of points, add 5, and call the result points*.

Python has a shorter way to write this. The line `points += 5` does the same job as `points = points + 5`. The signs `-=`, `*=` and `/=` work in the same way.

## 4. Four kinds of value

Every value in Python has a **type**. The type tells Python what kind of thing the value is. The function `type()` shows the type of a value. Write your guess for each line, and then run the cell.

````python
# My guess for each line:
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

Python has many types. These four appear in almost every program.

An **int** is a whole number, such as 42 or -7. A **float** is a number with a decimal point, such as 3.14. A **str**, short for *string*, is text between quote marks, such as `"hello"`. A **bool**, short for *boolean*, is either `True` or `False`.

Python chooses the type from the way you write the value. The decimal point makes a float, and the quote marks make a string.

**4.1** **Predict.** Write your guess for the type of each value, and then run the cell. Which one surprised you?

````python
# My guess for each line:
print(type(7))
print(type(7.0))
print(type("7"))
print(type(7 > 3))
print(type(10 / 2))
````

The number 7 is an int, and `7.0` is a float, because of the decimal point. The text `"7"` is a str, because of the quote marks, even though it looks like a number. A comparison such as `7 > 3` gives a bool. The sign `/` always gives a float, so `10 / 2` is `5.0` and not `5`.

The signs `+`, `-` and `*` add, subtract and multiply. The sign `/` divides and gives a float. The sign `**` raises a number to a power, so `7 ** 2` is 7 times 7.
