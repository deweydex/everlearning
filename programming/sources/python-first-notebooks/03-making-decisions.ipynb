## 1. Questions with a True or False answer

A decision starts with a question. In Python, a question about two values gives a bool: `True` or `False`. Six signs make the questions.

| Sign | The question |
|---|---|
| `==` | Are the two sides equal? |
| `!=` | Are the two sides different? |
| `>` | Is the left side bigger? |
| `<` | Is the left side smaller? |
| `>=` | Is the left side bigger than the right side, or equal to it? |
| `<=` | Is the left side smaller than the right side, or equal to it? |

Write your guess for each line, and then run the cell.

````python
# My guess for each line:
age = 17
print(age > 18)
print(age >= 17)
print(age == 17)
print(age != 17)
print(age < 17)
print(age <= 17)
````

Remember that `=` attaches a name to a value, and `==` asks a question. The two signs look alike, and they do different jobs.

**1.1** **Predict.** Write your guess for each line, and then run the cell. Which results surprised you?

````python
# My guess for each line:
print("apple" < "banana")
print("cat" == "cat ")
print("Zebra" < "apple")
print(3 == 3.0)
print("3" == 3)
````

Python compares strings letter by letter, in the order of the alphabet, so `"apple" < "banana"` is `True`. Every capital letter comes before every small letter in Python's order, so `"Zebra" < "apple"` is also `True`. A space counts as a character, so `"cat"` and `"cat "` are different. The int `3` and the float `3.0` have the same value, so they are equal. The string `"3"` and the int `3` are different types, so they are never equal.

When a program compares words that a person typed, the capital letters can cause trouble. The methods `.lower()` and `.strip()` from Notebook 1 help.

## 4. Many paths: elif

Some questions have more than two answers. The word `elif` means *else, if*. It adds another condition. Write your guess for the output with `mark = 72`, and then run the cell. Change the mark to 85, 50 and 20 and test each one.

````python
# My guess:
mark = 72
if mark >= 80:
    print("Distinction")
elif mark >= 65:
    print("Merit")
elif mark >= 50:
    print("Pass")
else:
    print("Unsuccessful")
````

Python checks the conditions from the top to the bottom. It runs the block of the first condition that is `True`, and then it skips everything else. For a mark of 72, the first condition is `False`, the second is `True`, so the program prints `Merit` and never looks at the third condition.

The order of the conditions matters. Predict the output of this cell before you run it.

````python
# My guess:
mark = 85
if mark >= 50:
    print("Pass")
elif mark >= 65:
    print("Merit")
elif mark >= 80:
    print("Distinction")
else:
    print("Unsuccessful")
````

The cell printed `Pass` for a mark of 85. The first condition, `mark >= 50`, is `True` for 85, so Python ran that block and skipped the others. The conditions for `Merit` and `Distinction` never had a chance.

**C7** **Is it a vowel?** The word `in` asks whether a short string sits inside a longer string. Run the cell. Then type a letter, type a capital letter, type two letters, and press Enter with nothing typed. What do you notice? Write a program that tells you whether a typed letter is a vowel, and test the strange cases.

````python
letter = input("A letter: ")
print(letter in "aeiou")
````

<details>
<summary>Hint</summary>

The empty string sits inside every string, so one case needs its own check. The method `.lower()` handles capital letters, and `len(letter) == 1` handles the other cases.

</details>

<details>
<summary>One answer</summary>

```python
letter = input("A letter: ").strip().lower()
if len(letter) != 1:
    print("Please type one letter.")
elif letter in "aeiou":
    print("vowel")
else:
    print("not a vowel")
```

Here is one answer. Yours may be different and work too.

Pressing Enter with nothing typed gives the empty string, and `"" in "aeiou"` is `True`. A check on the length stops that case before the question about vowels.

</details>

# Word list

| Word | What it means |
|---|---|
| condition | A question that has a bool as its answer. |
| comparison | A condition made with a sign such as `==`, `<` or `>=`. |
| `if` | Runs a block of lines when a condition is `True`. |
| `else` | Runs a block when the condition of the `if` is `False`. |
| `elif` | Another condition that Python checks when the earlier ones were `False`. |
| block | The indented lines under an `if`, `elif` or `else`. |
| indentation | The spaces at the start of a line. They show which lines belong together. |
| `and`, `or`, `not` | Words that build one bool from other bools. |
| nested | An `if` statement inside the block of another statement. |
