## 1. What print shows

The `print` function shows values on the screen. Write your guess for the output, line by line, and then run the cell.

````python
# My guess for the output:
print("Hello")
print("Hello", "world")
print("Hello" + "world")
print()
print("Goodbye")
````

The output has five lines. When `print` receives several values separated by commas, it puts a space between them. The sign `+` joins two strings with no space. The call `print()` with nothing inside shows an empty line. Each call to `print` ends by moving to a new line.

The function has options, written after the values. The option `sep` sets what goes between the values, and the option `end` sets what goes at the end of the output. Run the cell and read the four results.

````python
print("a", "b", "c", sep="-")
print("Loading", end="...")
print("done")
print("Line one\nLine two")
````

The first line shows `a-b-c`, because `sep="-"` replaced the spaces. The second line stays open, because `end="..."` replaced the new line, so `done` follows on the same line. The last call shows two lines. The two characters `\n` stand for a new line, and `\t` stands for a tab space.

## 2. Putting values inside a sentence

A program often needs a sentence with a value in the middle of it. Python has three ways to build the sentence. Write your guess for each line, and then run the cell.

````python
# My guess for each line:
total = 12.5
print("Total:", total)
print("Total: " + str(total))
print(f"Total: {total}")
````

All three lines show the same sentence. The first way uses a comma, and `print` adds the space. The second way changes the number into a string with `str()` and joins the two strings with `+`. You must write the space inside the quote marks.

The third way uses an **f-string**. An f-string is a string with the letter `f` before the first quote mark. Python looks for names inside curly brackets `{ }` and replaces each name with its value. An f-string lets us read the whole sentence in one piece, so the form is the one we use most.

The curly brackets can hold a calculation too. The line `print(f"Double: {total * 2}")` shows `Double: 25.0`.

Money needs two decimal places. An f-string can set the number of decimals. The part after the colon is a **format**, and `.2f` means *a float with two decimals*. Write your guess for each line, and then run the cell.

The format `.2f` shows exactly two digits after the decimal point, even for the whole number 3. The format rounds the number it shows, so `7 / 3` shows as `2.33`, and `2 / 3` with `.1f` shows as `0.7`. The variable itself keeps its full value. Only the displayed text changes.

These are the four names and what each one tells us.

| Error name | What it tells us |
|---|---|
| `NameError` | Python does not know this name. Look for a typo, a capital letter that does not match, or a variable that no cell has made yet. |
| `TypeError` | Python cannot do this operation with these types, such as joining a string and an int. |
| `ValueError` | The type is acceptable and the value is not, such as `int("hello")`. |
| `SyntaxError` | Python cannot read the shape of the line, for example when a quote mark or a bracket is missing. |

A `SyntaxError` appears before the cell runs at all, so nothing in the cell is shown. The other three errors appear when Python reaches the line, so the lines above the error have already run.

# Word list

| Word | What it means |
|---|---|
| input | A function that shows a prompt, waits for a typed line, and gives the line back as a string. |
| prompt | The message that `input` shows before it waits. |
| output | What a program shows on the screen. |
| f-string | A string that starts with `f` and holds names or calculations in curly brackets. |
