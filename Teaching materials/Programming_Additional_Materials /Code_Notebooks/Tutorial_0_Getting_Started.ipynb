## Your First ML-Themed Program! 🤖

Let's create a simple **prediction system** using a linear equation. This is the foundation of many ML algorithms!

### The Problem
Imagine we want to predict a student's exam score based on study hours using this simple equation:

**Score = (Study Hours × 10) + 20**

This is our "trained model" (expert system) based on observations.

```python
# Simple Prediction System
# This is like a very basic ML model!

# Input: study hours
study_hours = 5

# Our "model" - a mathematical equation
predicted_score = (study_hours * 10) + 20

# Output: prediction
print("=== Student Score Predictor ===")
print("Study Hours:", study_hours)
print("Predicted Score:", predicted_score)
print("================================")
```

### 🎯 Understanding the Code

1. **Input**: `study_hours = 5` - Our input data
2. **Model**: `(study_hours * 10) + 20` - Our prediction formula
3. **Output**: `predicted_score` - Our prediction

This is **exactly** how many ML systems work:
- Take input data
- Apply a mathematical model
- Produce a prediction
